## Question 1: Comprendre les KPIs


|Indicateur|Formule|
----------|-----
|CTR|le nombre de click * 100/ impression|
|CPC|le cout par  click = cout/le nbr de click|
|CPM|le cout sur 1000 impression = le cout total* 1000/ impression|
|CPL|cout sur lead = le cout/nbr de lead(qui contactent l'ecole)|
|Conversion rate|Leads*100 / Clics|
|ROAS|revenue/budget depense|

#### Question complémentaire : quelle est la différence entre un clic, un lead et une inscription payante ?  
**Reponse** : la difference reside dans les niveaux de l'entenoir marketing, d'ou le clic represente la phase de sensibilisation, le lead represente la consedireation et l'inscription devient la conversion.

## Question 2: Calculer les performances 

```python
import pandas as pd
df = pd.read_csv("ecol_compaign.csv")

# CTR (%)
df["CTR"] = df["clicks"]*100/df["impressions"]

# CPC (MAD)
df["CPC"] = df["spend_mad"]/df["clicks"]

# CPM(MAD)
df["CPM"] = df["spend_mad"]*1000/df["impressions"]

# CPL(MAD)
df["CPL"] = df["spend_mad"]/df["leads"]

# TAUX DE CONVERSION CLIC VS LEAD (%)
df["CLIC_LEAD"] = df["leads"]*100/df["clicks"]

# TAUX DE CONVERSION LEAD VS INSCRIPTION (%)
df["LEAD_INSCRIPTION"] = df["enrollments"]*100/df["leads"]

# ROAS
df["ROAS"] = df["revenue_mad"]/df["spend_mad"]

# AFFICHER LES RESULTATS
df
```

## Question 3:  SQL — extraire les performances

1. Afficher les donnees stockees dans un tableau SQL "campaigns":   

**SELECT**
    campaign, spend_mad,  
    ROUND(100.0 * clicks / impressions, 2) **AS** ctr,  
    ROUND(spend_mad / clicks, 2) **AS** cpc,  
    ROUND(spend_mad / leads, 2) **AS** cpl  
**FROM** campaigns;  

2. filtrer les campagnes dont le CPL est inférieur à 16 MAD :  

**SELECT** spend_mad/leads **AS** cpl FROM campagns   
**WHERE** spend_mad/leads<16;
